## Polymorphism & Inheritance

```python
class animal:
    def __init__(self,name):
        self.name=name
    def sound(self):
        print(f"{self.name} makes sound")

class dog(animal):
    def sound(self):
        print(f"{self.name} says woof")

class cat(animal):
    def sound(self):
        print(f"{self.name} says meow")

class cow(animal):
    def sound(self):
        print(f"{self.name} says Moo")

class Lion(animal):
    def sound(self):
        print(f"{self.name} roars")

        
```

```python
d=dog("Tommy the dog")
c=cat("kitty the cat")
cw=cow("Lakshmi the cow")
l=Lion("Simba the lion")
```

```python
animals=[d,c,cw,l]

for a in animals:
    a.sound()
```

## Abstraction and Inheritance

```python
from abc import ABC,abstractmethod

class animal(ABC):
    def __init__(self,name):
        self.name=name
    def sleep(self):
        print(f"{self.name} is sleeping")
    @abstractmethod
    def sound(self):
        pass

class dog(animal):
    def sound(self):
        print(f"{self.name} says woof")

class cat(animal):
    def sound(self):
        print(f"{self.name} says meow")

class cow(animal):
    def sound(self):
        print(f"{self.name} says Moo")

class Lion(animal):
    def sound(self):
        print(f"{self.name} roars")
        
```

```python
d2=dog("Tommy the dog")
c2=cat("Kitty the cat")
cl2=cow("Lakshmi the cow")
l2=Lion("Simbha the lion")
```

```python
d2.sound()
d2.sleep()
c2.sound()
c2.sleep()
cl2.sound()
cl2.sleep()
l2.sound()
l2.sleep()
```
