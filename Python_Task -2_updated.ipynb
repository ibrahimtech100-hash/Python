# 1.while loop 

### Print odd numbers upto 9 (eg-1)

```python
i = 1
while i <= 9:
    print(i)
    i = i + 2
```

### sum of N natural numbers (eg-2)

```python
n=int(input(" enter a number:"))
i=1
total=0

while i<=n:
    total=total+i;
    i=i+1;

print("sum of N natural numbers", total)
```

# 5.User defined function without for loop, if (while condition)

### Electricity bill slab (eg-1)

```python
def eb_slab():
    while True:
        units=int(input("Enter the eb units consumed (Enter 0 to stop)"))
        if units==0:
            print("Program stopped")
            break
        elif units <100:
            print("No need to pay bill.Less than 100 units is free")
        else:
            print("Need to pay bill")
eb_slab()
```

### Check leap year (eg-2)

```python
def leap_year():
    while True:
        year=int(input("enter a year (Enter 0 to stop)"))
        if year==0:
            print("program stopped")
            break
        elif year%4==0:
            print("This is leap year")
        else:
            print("Not a yeap lear")

leap_year()
```

# 6.User defined function with for loop (without if condition)

### Multiplication table (eg-1)

```python
def multiplication_table(n):
    for i in range(1,11):
        print(n,"x",i,"=",n*i)
multiplication_table(5)
```

### Reverse of number (eg-2)

```python
def reverse_number(n):
    for i in range(n,0,-1):
        print (i)

reverse_number(5)
```

# 7.User defined function with for loop with if condition

### Check how many even numbers in a range (eg-1)

```python
def even_number(number):
    count=0
    for i in range(1,number+1):
        if i%2==0:
           count=count+1;
    print("The total number of even numbers is",count)

even_number(30)
```

### Factors of a number (eg-2)

```python
def factors(csk):
    for ipl in range (1,csk+1):
        if csk%ipl==0:
          print(ipl)
factors(15)
```

### Print students not present (eg-1)

```python
all_students=["Sathya","Prakash","Moorthy","Ganesh","Raju","Anjali"]
present=["Sathya","Anjali","Ganesh"]

for i in all_students:
    if i not in present:
        print(i)
```
