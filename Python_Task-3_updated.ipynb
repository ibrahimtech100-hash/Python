```python

import random
def RPS_game():
    option=["rock","paper","scissor"]
    print("Welcome to Rock Paper Scissor Game")
    computer_option=random.choice(option)

    while True:
       computer_option=random.choice(option)
       user_input=input("Enter Your choice: Rock/Paper/Scissor").lower()

       if user_input not in option:
          print("Type valid options. Rock/Paper/Scissor")
          continue
       if user_input == computer_option:
          print(f"It is a tie. Both choose {user_input}")
       elif user_input =="rock": 
         if computer_option =="scissor":
           print("You win. Rock will broke scissor")
         else:
           print("You loose. Paper wraps rock")
       elif user_input =="paper": 
         if computer_option =="scissor":
           print("You loose. scissor cuts paper")
         else:
           print("You win. Paper will wrap rock")
       elif user_input =="scissor": 
         if computer_option =="paper":
           print("You win. scissor cuts paper")
         else:
           print("You loose. Rock will broke scissor")

       user_input_2=input("want to Play game again? - Yes/No").lower()
       if user_input_2 != "yes":
            print("Thanks for playing")
            break

    
RPS_game()
```
